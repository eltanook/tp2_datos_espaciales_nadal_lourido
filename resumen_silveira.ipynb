# Fundamentos Teóricos y Metodología de Teledetección

## 1. Introducción y Planteamiento del Problema
Bienvenidos a la primera sesión. Hoy analizamos el paper *"Spatio-temporal remotely sensed indices identify hotspots of biodiversity conservation concern"*. El objetivo central de los autores fue desarrollar índices basados en satélites para identificar **puntos calientes (hotspots)** donde la conservación de la biodiversidad es prioritaria.

El problema raíz que motiva este estudio es el **Cambio Climático**, específicamente cómo la variabilidad del clima amenaza la supervivencia de las especies. Sin embargo, no todos los lugares responden igual. Los autores proponen que la clave para priorizar la conservación está en la intersección de dos dimensiones: el **TIEMPO** (variabilidad interanual) y el **ESPACIO** (variabilidad espacial).

### A. La Teoría de la Amenaza: Desajustes Fenológicos
Primero, hablemos del tiempo.
* **El Concepto:** Las especies dependen de la sincronización (sincronía). Por ejemplo, un ave migratoria llega a un bosque esperando encontrar orugas para alimentar a sus crías.
*  **El Riesgo:** Si hay una alta **variabilidad interanual** (un año la primavera empieza en agosto, el otro en octubre), se produce un **desajuste fenológico**.
*  cia:** Si las especies no pueden adaptarse a estos cambios erráticos, se produce un desajuste entre la demanda de recursos (ej. cría) y la disponibilidad de estos (ej. brote de hojas o insectos), llevando al declive poblacional.
*  del Paper:** Áreas con alta variabilidad interanual en verdor o temperatura representan una **ALTA AMENAZA**.

### B. La Teoría de la Resiliencia: Refugios Espaciales
Ahora, miremos el espacio.
* **El Concepto:** La heterogeneidad del paisaje. Un terreno con montañas, valles y diferentes tipos de cobertura vegetal en un área pequeña.
*  cio:** Esta variabilidad espacial crea **microclimas y refugios**. Si hace demasiado calor en la ladera norte, un animal puede moverse a la ladera sur.
* **Supuesto del Paper:** Una alta variabilidad espacial aumenta la **resiliencia ecológica**.   las especies persistan incluso durante años climáticamente malos.

## 3. La "Caja de Herramientas" Satelital (Metodología)
Para medir estas variables teóricas desde el espacio, utilizaron dos sensores satelitales distintos, aprovechando las fortalezas de cada uno.

### A. Midiendo la Amenaza (Análisis Temporal 2001-2018)
Necesitaban ver la evolución en el tiempo con alta frecuencia.
* **Sensor:** **MODIS** (a bordo del satélite Terra).
*  MODIS?** Tiene una resolución temporal muy alta (imágenes diarias o cada pocos días), lo que permite construir curvas de crecimiento de la vegetación muy precisas.
* **Datos usados:**
    1.  **EVI (Índice de Vegetación Mejorado):** Mide el "verdor" (productividad).  VI sobre NDVI porque el EVI no se satura tanto en bosques densos y corrige mejor los efectos atmosféricos. Resolución: 250m.
    2.  **LST (Temperatura de Superficie Terrestre):** Mide el calor emitido por la tierra.  Resolución: 1km.
* **La Métrica Clave (CV):** No les importaba el promedio, sino la variabilidad.  el **Coeficiente de Variación (CV)** del "Inicio de la Temporada de Crecimiento" (SOS - Start of Season) a lo largo de 18 años.
    * *Traducción:* Si el CV es alto, significa que la primavera es impredecible (amenaza alta).

### B. Midiendo la Resiliencia (Análisis Espacial)
Necesitaban ver el detalle del terreno.
* **Sensor:** **Landsat 8**.
*  Landsat?** Tiene mejor resolución espacial (30 metros) que MODIS, permitiendo ver detalles finos del paisaje.
*  **La Métrica Clave (Textura de Imagen):** Utilizaron la **Desviación Estándar** calculada en una "ventana móvil" de $11\times11$ píxeles.
    * *Procedimiento:* El algoritmo mira un cuadrado de 11x11 píxeles, calcula qué tan diferentes son los valores entre sí (desviación estándar) y asigna ese valor al píxel central.
    *  *Interpretación:* Un valor alto de desviación estándar significa que en ese pequeño cuadrado hay mucha diversidad de condiciones (ej. sol, sombra, roca, agua), lo que indica **alta resiliencia/refugios**.

# Resultados, Diagnóstico Territorial y Acción Política

## Módulo 4: Resultados Clave (El Diagnóstico Satelital)

Al procesar 18 años de datos MODIS y miles de imágenes Landsat, los autores encontraron patrones que desafían algunas suposiciones comunes en ecología.

### 1. El Gran Desacople: Verdor vs. Temperatura
Solemos pensar que si la temperatura se vuelve loca, la vegetación también lo hará inmediatamente. El estudio demostró que esto **no es lineal**.
* **El Hallazgo:** La correlación entre la variabilidad del inicio de la temporada de verdor (EVI) y la variabilidad de la temperatura (LST) fue sorprendentemente débil e incluso negativa a nivel nacional ($r = -0.19$).
* **La Lección:** No podemos usar la temperatura como único indicador. cies responden al calor (exotermos), pero otras (herbívoros) dependen del verdor. vo, **debemos medir ambas variables por separado**, ya que ocurren en lugares diferentes.

### 2. Identificación de Hotspots (Puntos Calientes)
los mapas de amenaza y resiliencia, Argentina quedó dividida en zonas claras de preocupación:

#### A. La "Zona Roja": Máxima Preocupación de Conservación
* **Ubicación:** Noreste de Argentina. Predominantemente en las ecorregiones del **Chaco Húmedo**, **Espinal** y **Esteros del Iberá**.
*  Afectadas:** Corrientes, Santa Fe y Entre Ríos tienen más del 50% de su territorio en esta categoría.
* **El Diagnóstico:** Estas áreas enfrentan una "tormenta perfecta":
    1.  lidad interanual (el clima y la fenología son impredecibles).
    2.  lidad espacial (son llanuras homogéneas, sin refugios topográficos naturales).
* **El Riesgo:** Si ocurre un evento climático extremo, las especies no tienen a dónde ir "aquí al lado" para salvarse.

#### B. La "Zona Verde": Baja Preocupación (Alta Resiliencia)
* **Ubicación:** Noroeste y Centro-Oeste.  de **Yungas**, **Bosques Patagónicos**, **Puna** y **Altos Andes**.
* **El Diagnóstico:**
    1.  Baja variabilidad interanual (patrones temporales estables).
    2.  lidad espacial (gran diversidad de relieves y microclimas).
* **El Valor:** Estas áreas actúan como **bancos de seguridad** biológica. La complejidad del terreno permite que las especies encuentren nichos térmicos adecuados incluso si el clima global cambia.

## Módulo 5: Implicaciones de Manejo (¿Qué hacemos con esta información?)

El paper no solo diagnostica, sino que prescribe tratamientos específicos para cada zona. Las estrategias deben ser radicalmente opuestas dependiendo de si estamos en la "Zona Roja" o la "Zona Verde".

### Estrategia para la "Zona Roja" (Noreste: Chaco, Espinal, Iberá)
Aquí la resiliencia natural es baja, por lo que la intervención humana es urgente para evitar colapsos de biodiversidad.
* **El Problema:** Estas zonas ya están muy degradadas por la agricultura y ganadería, y tienen pocas áreas protegidas (alrededor del 10%).
* **La Solución Propuesta:**
    1.  ón de Conectividad:** Crear corredores que conecten los pocos refugios que quedan.
    2.  Humedales:** Mantener la conectividad entre humedales y fuentes de agua abiertas es vital para asegurar recursos alimenticios durante sequías.
    3.  **Aumentar la Heterogeneidad:** "Desordenar" el paisaje productivo. ión de vegetación nativa en matrices agrícolas puede crear la variabilidad espacial que la naturaleza no proveyó geológicamente.

### Estrategia para la "Zona Verde" (Yungas y Patagonia)
Aquí la naturaleza nos ha regalado resiliencia. La estrategia es **blindar** estas áreas.
* culto:** Estas áreas pueden sostener la biodiversidad a largo plazo y mantener la capacidad adaptativa ante el cambio climático.

## Módulo 6: Crítica Política y la Ley de Bosques

Este es quizás el punto más contundente del paper. Los autores cruzaron sus mapas científicos con el mapa político de la **Ley de Bosques Nativos (N° 26.331)** de Argentina y encontraron una discrepancia peligrosa.

### El Semáforo de la Ley de Bosques
La ley clasifica los bosques en:
* **Zona I (Rojo):** Valor de conservación muy alto. No se toca.
* **Zona II (Amarillo):** Valor medio. sostenible (silvopastoreo, turismo, recolección).
* **Zona III (Verde):** Valor bajo. ricultura.

### La Denuncia del Paper
El estudio encontró que en los **hotspots de alta resiliencia** (noroeste y centro-oeste), la protección es insuficiente:
* El **58%** de estos bosques resilientes están clasificados actualmente como **Zona II (Amarillo)**.
* tá en **Zona III (Verde)**.
* Solo el **36%** está protegido bajo **Zona I (Rojo)**.

### La Recomendación Política
Los autores argumentan que permitir actividades productivas (Zona II) o desmonte (Zona III) en estos refugios climáticos es un error estratégico.
* **Propuesta:** Los bosques actualmente en Zona II y III que caen dentro de estos "hotspots de resiliencia" deberían ser recategorizados urgentemente a **Zona I**.
* **Justificación:** Dado su alto valor potencial para amortiguar el cambio climático, su preservación estricta es más valiosa para la nación que su explotación productiva.

## Conclusión Final de la Clase

El trabajo de Silveira et al. (2021) nos enseña una lección fundamental para la era del Big Data ambiental:
1.  **Mirar más allá del promedio:** No basta con saber la temperatura media; la **variabilidad** es lo que mata o salva a las especies.
2.  **La tecnología guía la política:** Los satélites nos muestran que estamos protegiendo insuficientemente nuestras áreas más resilientes (Yungas/Patagonia) y descuidando las más amenazadas (Chaco/Espinal).
